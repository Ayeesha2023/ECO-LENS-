```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/ECO-LENS/Master_Set/YOLO_Outputs/eco_lens_final/weights/best.pt")

print("✅ ECO-LENS model loaded successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
image_path = "/content/drive/MyDrive/ECO-LENS/Taco/test/images/000004_jpg.rf.9da65fb141388e0b5f851e4847eb594a.jpg"
```

```python
results = model.predict(
    source=image_path,
    conf=0.25,
    save=True
)
```

```python
import matplotlib.pyplot as plt

result = results[0]

plt.figure(figsize=(10,10))
plt.imshow(result.plot())
plt.axis("off")
plt.show()
```

```python
for box in results[0].boxes:
    cls = int(box.cls)
    conf = float(box.conf)

    print(f"Class: {model.names[cls]}")
    print(f"Confidence: {conf:.2f}")
    print("---------------------")
```
